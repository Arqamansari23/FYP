```python
from langchain.document_loaders import UnstructuredURLLoader
```

```python
loader = UnstructuredURLLoader(
    urls = [
        "https://economictimes.indiatimes.com/itc-ltd/stocks/companyid-13554.cms#financials",
        # "https://www.moneycontrol.com/news/business/markets/market-corrects-post-rbi-ups-inflation-forecast-icrr-bet-on-these-top-10-rate-sensitive-stocks-ideas-11142611.html"
    ]
)
```

```python
data = loader.load()
data
```

```python
from langchain.text_splitter import RecursiveCharacterTextSplitter

r_splitter = RecursiveCharacterTextSplitter(
    separators = ["\n\n", "\n", " "],  # List of separators based on requirement (defaults to ["\n\n", "\n", " "])
    chunk_size = 1000,  # size of each chunk created
    chunk_overlap  = 300,  # size of  overlap between chunks in order to maintain the context
    

)
```

```python
chunks = r_splitter.split_documents(data)
```

```python

from langchain_community.embeddings import HuggingFaceEmbeddings

huggingface_embeddings=HuggingFaceEmbeddings(
    model_name="BAAI/bge-small-en-v1.5",      #sentence-transformers/all-MiniLM-l6-v2
    model_kwargs={'device':'cpu'},
    encode_kwargs={'normalize_embeddings':True}

)
```

```python
query_result=huggingface_embeddings.embed_query("How are You ?")
len(query_result)
```

```python
# Loop through each chunk
for chunk in chunks:
  # Get the text content of the chunk
  text_content = chunk.page_content

  # Encode the text content into an embedding using HuggingFaceBgeEmbeddings
  embedding = huggingface_embeddings.encode(text_content)

  # Add the document (text content and embedding) to the Pinecone retriever
  retriver.add_document(text_content, embedding)
```
